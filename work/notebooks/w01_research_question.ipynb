## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, pandas as pd

candidates = [
  "../../data/raw/content_refresh_anonymized.csv"
]
path = next((p for p in candidates if os.path.exists(p)), None)

df = pd.read_csv(path, low_memory=False)
total = len(df)
high_vol = int((df.get("impressions_90d", pd.Series(dtype=float)) >= 500).sum())
down = int((df.get("trend_direction", pd.Series(dtype=str)) == "down").sum())

print(f"Total rows: {total}")
print(f"Rows with impressions_90d >= 500: {high_vol}")
print(f"Rows with trend_direction == 'down': {down}\n")

print("Total rows: {total}; Rows with impressions_90d ≥ 500: {high_vol}; Rows with trend_direction == 'down': {down} — enough signal and volume to build and validate a temporal prediction model.".format(total=total, high_vol=high_vol, down=down))
```
